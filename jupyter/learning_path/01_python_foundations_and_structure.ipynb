**Goal:** understand the Python ideas that make a multi-file project feel manageable.

```text
input → function transforms it → returned output
```

A file should own one related responsibility. A function should have clear inputs and outputs.

## Functions, modules, and imports

- A **function** packages one operation.
- A **module** is a Python file containing related functions/classes.
- An **import** lets one module use another module's public capability.
- `if __name__ == "__main__":` prevents a script from running its interactive code when another file imports it.

Comments should explain **why** a decision exists. Docstrings explain what a public function owns.

````python
def normalize_question(question: str) -> str:
    """Return one clean question before routing or retrieval."""
    cleaned = question.strip()
    if not cleaned:
        raise ValueError("Question cannot be empty")
    return cleaned


print(normalize_question("  What is RAG?  "))
````

## Dataclasses, type hints, and errors

A `dataclass` is useful when several named values belong together. Type hints describe the expected shape; they do not replace runtime validation. A custom exception gives one failure a meaningful application name.

````python
from dataclasses import dataclass


class InvalidBalanceError(Exception):
    pass


@dataclass
class Account:
    owner: str
    balance: int = 0

    def withdraw(self, amount: int) -> int:
        if amount > self.balance:
            raise InvalidBalanceError("Insufficient balance")
        self.balance -= amount
        return self.balance


account = Account("Harshu", 500)
print(account.withdraw(200))
````

## Data structures and Big-O

| Structure | Best use | Typical membership lookup |
|---|---|---|
| list | ordered sequence, duplicates allowed | O(n) |
| set | unique values, fast membership | O(1) average |
| dict | key → value lookup | O(1) average |
| deque | queue operations at either end | O(1) |

Big-O describes growth as input grows. It is not an exact wall-clock time.

````python
def count_nested_operations(size: int) -> int:
    operations = 0
    for _ in range(size):
        for _ in range(size):
            operations += 1
    return operations


for size in (10, 20, 40):
    print(size, count_nested_operations(size))
````
